# Microdata choice training: Web Appendix slope tables (`Tables A2–A6`) and diagnostics

Full attribute-level slope tables for `Vaccines` (`Table A2`), `Flights` (`Table A3`), `Consumer rights` (`Table A4`), `AI & offshoring` (`Table A5`), and `Food labels` (`Table A6`), plus the `AI & offshoring` country × arm × attribute-type diagnostic.

```python
import sys
from pathlib import Path
from IPython.display import display

PAPER_DIR = next(
    (
        p
        for p in (
            Path.cwd(),
            *Path.cwd().parents,
            Path.cwd() / "train-notebooks" / "beta-training" / "ms_paper_beta_loss",
            Path.cwd() / "code" / "train-notebooks" / "beta-training" / "ms_paper_beta_loss",
        )
        if (p / "microdata_analysis.py").is_file()
    ),
    None,
)
if PAPER_DIR is None:
    raise FileNotFoundError(f"Could not locate microdata_analysis.py from {Path.cwd()}")
if str(PAPER_DIR) not in sys.path:
    sys.path.insert(0, str(PAPER_DIR))

import microdata_analysis as ma

ctx = ma.load_microdata_context()
print(f"Loaded {len(ctx.regimes_by_model)} models from {ma.PAPER_RUNS_ROOT}")
```

```python
import numpy as np
import pandas as pd

for model_index, model_slug in enumerate(ma.MODEL_SLUGS):
    betas_by_slug = ctx.covid_betas_by_model[model_slug]
    columns = {"Human (OLS)": betas_by_slug[ma.BASE_SLUG]["human_pop_beta"]}
    for slug in ma.ALL_SLUGS:
        if slug in betas_by_slug:
            columns[ma.STORY_LABELS[slug]] = betas_by_slug[slug]["model_pop_beta"]
        else:
            columns[ma.STORY_LABELS[slug]] = pd.Series(np.nan, index=columns["Human (OLS)"].index)
    kreps_term_df = pd.DataFrame(columns)
    kreps_term_df.index.name = "Vaccine attribute level"
    squared_error = kreps_term_df.sub(kreps_term_df["Human (OLS)"], axis=0).pow(2)
    kreps_term_df.loc["β MSE vs. human"] = squared_error.mean()
    ma.print_caption(f"Table A2{ma.TABLE_LETTERS[model_index]}", f"Vaccines population β by attribute: {ma.MODEL_DISPLAY[model_slug]}")
    display(kreps_term_df.round(4))
    ma.print_note("Vaccines (Kreps et al., 2020) is a held-out test set. Human (OLS) = slope fitted on respondents' choices. The last row is β MSE versus human.")
```

```python
import pandas as pd

for model_index, model_slug in enumerate(ma.MODEL_SLUGS):
    betas_by_slug = ctx.kriner_betas_by_model[model_slug]
    columns = {"Human (OLS)": betas_by_slug[ma.BASE_SLUG].loc[ctx.kriner_terms, "human_pop_beta"]}
    for slug in ma.ALL_SLUGS:
        columns[ma.STORY_LABELS[slug]] = betas_by_slug[slug].loc[ctx.kriner_terms, "model_pop_beta"]
    kriner_term_df = pd.DataFrame(columns)
    kriner_term_df.index.name = "Flight attribute level"
    squared_error = kriner_term_df.sub(kriner_term_df["Human (OLS)"], axis=0).pow(2)
    kriner_term_df.loc["β MSE vs. human"] = squared_error.mean()
    ma.print_caption(f"Table A3{ma.TABLE_LETTERS[model_index]}", f"Flights population β by attribute: {ma.MODEL_DISPLAY[model_slug]}")
    display(kriner_term_df.round(4))
    ma.print_note("Values behind Figure 2a (Price and sustainable aviation fuel listed first). Human (OLS) = slope fitted on respondents' choices. Flights (Goldfarb et al., 2026) is the validation set and selects each run's checkpoint (best.pt).")
```

```python
import numpy as np
import pandas as pd

k26_human_reference = None
for model_slug in ma.MODEL_SLUGS:
    if ma.BASE_SLUG in ctx.kantorowicz_betas_by_model[model_slug]:
        ref_df = ctx.kantorowicz_betas_by_model[model_slug][ma.BASE_SLUG][["model", "human_pop_beta"]]
        if k26_human_reference is None:
            k26_human_reference = ref_df
        else:
            if list(ref_df.index) != list(k26_human_reference.index):
                raise ValueError(f"{model_slug}: Consumer rights term index mismatch")
            max_diff = float((ref_df["human_pop_beta"] - k26_human_reference["human_pop_beta"]).abs().max())
            if max_diff > 1e-12:
                raise ValueError(f"{model_slug}: Consumer rights human_pop_beta mismatch (max_diff={max_diff})")
if k26_human_reference is None:
    raise FileNotFoundError("No model has kantorowicz2026 in test_best/ yet")

k26_price_mask = pd.Series(
    [t.split(" | ", 1)[1].split(":", 1)[0] == "Price" for t in k26_human_reference.index],
    index=k26_human_reference.index,
)

for model_index, model_slug in enumerate(ma.MODEL_SLUGS):
    betas_by_slug = ctx.kantorowicz_betas_by_model[model_slug]
    columns = {"Human (OLS)": k26_human_reference["human_pop_beta"]}
    for slug in ma.ALL_SLUGS:
        if slug in betas_by_slug:
            columns[ma.STORY_LABELS[slug]] = betas_by_slug[slug].loc[k26_human_reference.index, "model_pop_beta"]
        else:
            columns[ma.STORY_LABELS[slug]] = pd.Series(np.nan, index=k26_human_reference.index)
    k26_term_df = pd.DataFrame(columns)
    k26_term_df.index.name = "Consumer rights attribute level"
    squared_error = k26_term_df.sub(k26_term_df["Human (OLS)"], axis=0).pow(2)
    k26_term_df.loc["Price β MSE vs. human (4)"] = squared_error.loc[k26_price_mask].mean()
    k26_term_df.loc["Non-price β MSE vs. human (34)"] = squared_error.loc[~k26_price_mask].mean()
    k26_term_df.loc["Overall β MSE vs. human (38)"] = squared_error.mean()
    ma.print_caption(f"Table A4{ma.TABLE_LETTERS[model_index]}", f"Consumer rights population β by attribute: {ma.MODEL_DISPLAY[model_slug]}")
    display(k26_term_df.round(4))
    ma.print_note(
        "Consumer rights (Kantorowicz-Reznichenko et al., 2026) is a held-out test set spanning 19 laptop and 19 laundry-machine AMCEs (38 total). "
        "Human (OLS) = slope fitted on respondents' choices. Summary rows report Price (4), Non-price (34), and Overall (38) β MSE versus human."
    )
```

```python
import numpy as np
import pandas as pd
from beta_ols import magistro2025 as ds_magistro
from beta_ols.core import fit_human_ols

m25_choice_df = ds_magistro.load_choice_df()
m25_rows = ds_magistro.regression_rows(m25_choice_df)
m25_ols = fit_human_ols(m25_rows, ds_magistro.regression_specs(m25_rows))
m25_human_reference = m25_ols.set_index("term")[["model", "human_beta"]].rename(
    columns={"human_beta": "human_pop_beta"}
)

for model_slug in ma.MODEL_SLUGS:
    if ma.BASE_SLUG in ctx.magistro_betas_by_model[model_slug]:
        ref_df = ctx.magistro_betas_by_model[model_slug][ma.BASE_SLUG][["model", "human_pop_beta"]]
        if list(ref_df.index) != list(m25_human_reference.index):
            raise ValueError(f"{model_slug}: AI & offshoring term index mismatch")
        max_diff = float((ref_df["human_pop_beta"] - m25_human_reference["human_pop_beta"]).abs().max())
        if max_diff > 1e-12:
            raise ValueError(f"{model_slug}: AI & offshoring human_pop_beta mismatch (max_diff={max_diff})")

m25_price_mask = pd.Series(
    [t.split(" | ", 1)[1].split(":", 1)[0] == "Price" for t in m25_human_reference.index],
    index=m25_human_reference.index,
)

for model_index, model_slug in enumerate(ma.MODEL_SLUGS):
    betas_by_slug = ctx.magistro_betas_by_model[model_slug]
    columns = {"Human (OLS)": m25_human_reference["human_pop_beta"]}
    for slug in ma.ALL_SLUGS:
        if slug in betas_by_slug:
            columns[ma.STORY_LABELS[slug]] = betas_by_slug[slug].loc[m25_human_reference.index, "model_pop_beta"]
        else:
            columns[ma.STORY_LABELS[slug]] = pd.Series(np.nan, index=m25_human_reference.index)
    m25_term_df = pd.DataFrame(columns)
    m25_term_df.index.name = "AI & offshoring attribute level"
    squared_error = m25_term_df.sub(m25_term_df["Human (OLS)"], axis=0).pow(2)
    m25_term_df.loc["Price β MSE vs. human (8)"] = squared_error.loc[m25_price_mask].mean()
    m25_term_df.loc["Non-price β MSE vs. human (24)"] = squared_error.loc[~m25_price_mask].mean()
    m25_term_df.loc["Overall β MSE vs. human (32)"] = squared_error.mean()
    ma.print_caption(f"Table A5{ma.TABLE_LETTERS[model_index]}", f"AI & offshoring population β by attribute: {ma.MODEL_DISPLAY[model_slug]}")
    display(m25_term_df.round(4))
    ma.print_note(
        "AI & offshoring (Magistro et al., 2025) is a held-out test set spanning 4 country × arm regressions of 8 AMCEs each (32 total). "
        "Human (OLS) = slope fitted on respondents' choices. Summary rows report Price (8), Non-price (24), and Overall (32) β MSE versus human."
    )
```

```python
import pandas as pd

M25_SUBSET_SPECS = (
    ("All (32)", lambda df: pd.Series(True, index=df.index)),
    ("US (16)", lambda df: df["model"].isin(["magistro2025/us_ai", "magistro2025/us_offshoring"])),
    ("Canada (16)", lambda df: df["model"].isin(["magistro2025/ca_ai", "magistro2025/ca_offshoring"])),
    ("US AI (8)", lambda df: df["model"] == "magistro2025/us_ai"),
    ("US Offshoring (8)", lambda df: df["model"] == "magistro2025/us_offshoring"),
    ("Canada AI (8)", lambda df: df["model"] == "magistro2025/ca_ai"),
    ("Canada Offshoring (8)", lambda df: df["model"] == "magistro2025/ca_offshoring"),
    ("US Price (4)", lambda df: df["model"].isin(["magistro2025/us_ai", "magistro2025/us_offshoring"]) & df.index.str.contains(r"\| Price:", regex=True)),
    ("Canada Price (4)", lambda df: df["model"].isin(["magistro2025/ca_ai", "magistro2025/ca_offshoring"]) & df.index.str.contains(r"\| Price:", regex=True)),
    ("US Jobs (12)", lambda df: df["model"].isin(["magistro2025/us_ai", "magistro2025/us_offshoring"]) & ~df.index.str.contains(r"\| Price:", regex=True)),
    ("Canada Jobs (12)", lambda df: df["model"].isin(["magistro2025/ca_ai", "magistro2025/ca_offshoring"]) & ~df.index.str.contains(r"\| Price:", regex=True)),
)

m25_subset_frames = []
for model_slug in ma.MODEL_SLUGS:
    betas_by_slug = ctx.magistro_betas_by_model[model_slug]
    rows = []
    for slug in ma.ALL_SLUGS:
        if slug not in betas_by_slug:
            raise KeyError(f"{model_slug}/{slug}: missing magistro2025 betas")
        beta_df = betas_by_slug[slug]
        sq_err = (beta_df["model_pop_beta"] - beta_df["human_pop_beta"]).pow(2)
        row = {}
        for col_label, mask_fn in M25_SUBSET_SPECS:
            mask = mask_fn(beta_df)
            if not mask.any():
                raise ValueError(f"{model_slug}/{slug}: empty subset for {col_label!r}")
            row[col_label] = float(sq_err.loc[mask].mean())
        rows.append(row)
    frame = pd.DataFrame(rows)
    frame.index = pd.MultiIndex.from_tuples(
        [(ma.MODEL_DISPLAY[model_slug], slug) for slug in ma.ALL_SLUGS],
        names=["Model", "Trained on"],
    )
    m25_subset_frames.append(frame)

m25_subset_values = pd.concat(m25_subset_frames)
m25_subset_text = pd.DataFrame("", index=m25_subset_values.index, columns=m25_subset_values.columns)
for model_name, slug in m25_subset_values.index:
    for col in m25_subset_values.columns:
        val = float(m25_subset_values.loc[(model_name, slug), col])
        base_val = float(m25_subset_values.loc[(model_name, ma.BASE_SLUG), col])
        m25_subset_text.loc[(model_name, slug), col] = ma.format_mse_cell(val, base_val, slug)

m25_subset_css = ma.relabel_rows(ma.best_cell_css(m25_subset_values))
ma.print_caption("Diagnostic table", "AI & offshoring population β MSE (×10³) by country, arm, and price vs. job attributes")
display(ma.relabel_rows(m25_subset_text).style.apply(lambda _df: m25_subset_css, axis=None))
ma.print_note(
    "Values are population β MSE ×10³ on AI & offshoring (Magistro et al., 2025), with % change versus Off-the-shelf in parentheses; lower is better. "
    "Bold marks the lowest trained value per model and column when it improves on Off-the-shelf."
)
```
